```python
import pandas as pd
from pathlib import Path
```

```python
LME_DATA = Path.cwd() / "data" / "lme_aluminium.csv"
MW_PREMIUM_DATA = Path.cwd() / "data" / "midwest_premium.csv"
US_IMPORTS_DATA = Path.cwd() / "data" / "us_imports_by_country.csv"
```

```python
lme = pd.read_csv(LME_DATA)
mw = pd.read_csv(MW_PREMIUM_DATA)
us_imports = pd.read_csv(US_IMPORTS_DATA)
```

```python
mw.columns
```

```python
# checking whether monthly average for LME prices are the same for both data sources
# mw data starts in 2020-08-01
start_date = '2020-08-01'
monthly = lme.groupby(
    pd.to_datetime(lme.loc[lme['date'] > start_date]['date']
                   ).dt.to_period('M'))['cash'].mean()
```

```python
differences = monthly.reset_index(drop=True) - mw['lme_cash_usd_t'].reset_index(drop=True)
print(differences.mean(), 
    differences.std(), 
    differences.mean()/monthly.mean()) # some disagreement but not relatively important
```
